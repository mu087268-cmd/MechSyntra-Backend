```python
import pandas as pd
import numpy as np

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import mean_squared_error

print("Libraries Imported Successfully")

# Load Dataset

data = pd.read_csv("zoo.csv")

print("\nFirst 5 Rows of Dataset:\n")

print(data.head())

# Dataset Shape

print("\nDataset Shape:")

print(data.shape)

# Column Names

print("\nColumn Names:")

print(data.columns)

# Features and Labels

animal_names = data.iloc[:, 0]

X = data.iloc[:, 1:-1]

y = data.iloc[:, -1]

print("\nFeature Data:\n")

print(X.head())

# Find Unique Classes

unique_classes = y.nunique()

print("\nTotal Unique Classes =", unique_classes)

# Apply Agglomerative Clustering

model = AgglomerativeClustering(
    n_clusters=unique_classes,
    linkage='ward'
)

predicted_clusters = model.fit_predict(X)

print("\nPredicted Cluster Labels:\n")

print(predicted_clusters)

# Add Predicted Clusters

data['Predicted_Cluster'] = predicted_clusters

print("\nAnimal Grouping Results:\n")

print(data[['animal_name', 'class_type', 'Predicted_Cluster']].head(20))

# Calculate MSE

mse = mean_squared_error(y, predicted_clusters)

print("\nMean Squared Error =", mse)

# Final Conclusion

print("\nAgglomerative Clustering applied successfully.")

print("Assignment Completed Successfully")
```
